```python
def to_prompt(ds):
    return {
        "prompt": f"Contdst: {ds['context']}\nQuestion: {ds['question']}\n"
                  f"A. {ds['ans0']}\nB. {ds['ans1']}\nC. {ds['ans2']}\n"
                  f"Answer with A, B, or C only."
    }

ds_p = ds.map(to_prompt)
print(ds_p[0]["prompt"])
```
